```python
#getting the ground truth
app = 'dish washer'

ground_truth = results['gt'][app].values

ground_truth_vec =  results['gt']

mains = results['test_mains'][0].values.flatten()

#getting the predictions
predicted_values = results['predictions']['bert101'][app].values

predicted_values_vec =  results['predictions']['bert101']

#MAE function

def MAE(GT, predictions):
    return (1/len(GT))*sum([abs(GT[i] - predictions[i]) for i in range(0, len(GT))]);

def MR(GT, predictions):
    return sum([min(GT[i], predictions[i]) for i in range(0, len(GT))])/sum([max(GT[i], predictions[i]) for i in range(0, len(GT))]);

def NDE(GT, predictions):
    return math.sqrt(sum([(GT[i]-predictions[i])**2 for i in range(0, len(GT))])/sum([(GT[i])**2 for i in range(0, len(GT))]));

def NDE_overall(GT_vec, predictions_vec):
    return math.sqrt(sum([(GT_vec[i].values[j] - predictions_vec[i].values[j])**2 for i in GT_vec.keys() for j in range(0, len(ground_truth))])/sum([(GT_vec[i].values[j])**2 for i in GT_vec.keys() for j in range(0, len(ground_truth))]));


def SAE(GT, predictions):
    return abs(sum([predictions[i] for i in range(0, len(predictions))]) - sum([GT[i] for i in range(0, len(GT))]))/ sum([GT[i] for i in range(0, len(GT))]);

def SAE_overall(GT_vec, predictions_vec):
    return (1/len(GT_vec.keys()))*sum([ abs(sum([predictions_vec[i].values[j] for j in range(0, len(GT_vec[i].values))]) - sum([GT_vec[i].values[j] for j in range(0, len(GT_vec[i].values))]))/ sum([GT_vec[i].values[j] for j in range(0, len(GT_vec[i].values))]) for  i in GT_vec.keys()]);

def TECA(GT, predictions):
    return 1- (( sum([abs(GT[i] - predictions[i]) for i in range(0, len(GT))]) ) / (2*sum([GT[i] for i in range(0, len(GT))])));

def TECA_1(GT, predictions):
    return ( sum([abs(GT[i] - predictions[i]) for i in range(0, len(GT))]));

def TECA_2(GT, predictions):
    return sum([GT[i] for i in range(0, len(GT))]);
            
            
def TECA_overall(GT_vec, predictions_vec):
    return 1 - ( sum([abs(GT_vec[i].values[j] - predictions_vec[i].values[j]) for i in GT_vec.keys() 
                      for j in range(0, len(GT_vec[i].values))])) / (2*sum([mains[i] for i in range(0, len(mains))]));

def NEP(GT, predictions):
    return sum([abs(GT[i] - predictions[i]) for i in range(0, len(GT))]) / sum([GT[i] for i in range(0, len(GT))]);

def EA(GT, predictions, alpha): 
    return math.exp(-alpha*NEP(GT, predictions));

def NRMSE_range(GT, predictions):
    return math.sqrt((1/len(GT))*sum([(GT[i]-predictions[i])**2 for i in range(0, len(GT))])) / (max(GT) - min(GT));

def NRMSE_mean(GT, predictions):
    return math.sqrt((1/len(GT))*sum([(GT[i]-predictions[i])**2 for i in range(0, len(GT))])) / np.mean(GT);

def FEE(GT, predictions):
    return sum([predictions[i] for i in range(0, len(GT))])/sum([GT[i] for i in range(0, len(GT))]);

def CEP(GT, predictions):
    C = sum([predictions[i] for i in range(0, len(GT)) if predictions[i] == GT[i]])
    Cue = sum([predictions[i] for i in range(0, len(GT)) if predictions[i] < GT[i]])
    Coe = sum([GT[i] for i in range(0, len(GT)) if predictions[i] > GT[i]])
    return (C + Cue + Coe)/sum([GT[i] for i in range(0, len(GT))]);

def CEPmod(GT, predictions, alpha):
    C = sum([predictions[i] for i in range(0, len(GT)) if predictions[i] == GT[i]])
    Cue = sum([predictions[i] for i in range(0, len(GT)) if predictions[i] < GT[i]])
    Coe = sum([GT[i] for i in range(0, len(GT)) if predictions[i] > GT[i]])
    return (C + alpha*Cue + (1-alpha)*Coe)/sum([GT[i] for i in range(0, len(GT))]);

def FEEmod(GT, predictions, alpha):
    A = sum([predictions[i] for i in range(0, len(GT)) if predictions[i] == GT[i]])
    B = sum([predictions[i] for i in range(0, len(GT)) if predictions[i] < GT[i]])
    C = sum([predictions[i] for i in range(0, len(GT)) if predictions[i] > GT[i]])
    return (A + alpha*B + (1-alpha)*C)/sum([GT[i] for i in range(0, len(GT))]);
```
